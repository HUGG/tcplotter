# Installation

You can install T<sub>c</sub>plotter using `pip`.

```bash
pip install tcplotter
```

This also installs [T<sub>c</sub>ages](https://github.com/HUGG/tcages), which provides the thermochronometer age prediction codes used by T<sub>c</sub>plotter. Pre-built packages are available for Linux, macOS and Windows, so no compilers are needed.

**Note**: T<sub>c</sub>plotter versions up to 0.4.0 required installing separate age prediction programs. This is no longer necessary from version 0.5.0 onwards.

## Installing the latest version of T<sub>c</sub>plotter from GitHub

You can install the latest development version of T<sub>c</sub>plotter directly from GitHub using `pip`:

```bash
pip install git+https://github.com/HUGG/tcplotter.git
```

T<sub>c</sub>plotter requires Python >= 3.10. All required Python libraries (`matplotlib`, `numpy`, `scipy` and `tcages`) are installed automatically.

If you would like to modify the code, you can instead clone the repository and do an editable install:

```bash
git clone https://github.com/HUGG/tcplotter.git
cd tcplotter
pip install -e .
```

### Testing the installation

At this point you should have a working T<sub>c</sub>plotter installation. You can test it by running one of the T<sub>c</sub>plotter command-line programs. For example, you could run the following in a terminal where T<sub>c</sub>plotter is installed:

```bash
eu-vs-radius
```

This should produce a contour plot of cooling ages and closure temperatures for a range of effective uranium and grain radius values.

## Optional graphical progress bar

It is possible to enable a graphical progress bar with tcplotter when using it in a Jupyter notebook. This requires the [ipywidgets library](https://ipywidgets.readthedocs.io/). ipywidgets can also be installed using `pip` or `conda`:

```bash
pip install ipywidgets
```

or

```bash
conda install -c conda-forge ipywidgets
```
